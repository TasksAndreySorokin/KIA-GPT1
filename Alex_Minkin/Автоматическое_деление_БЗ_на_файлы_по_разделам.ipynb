```python
def split_file(input_file):
    with open(input_file, 'r') as file:
        lines = file.readlines()

    sections = []
    current_section = ''
    for line in lines:
        if line.startswith('# '):  # Начало нового раздела
            if current_section:
                sections.append(current_section)
            current_section = line
        else:
            current_section += line

    sections.append(current_section)  # Добавляем последний раздел

    for section in sections:
        section_name = section.split('\n', 1)[0][2:]  # Выделяем название раздела
        output_file = f"{section_name}.md"

        with open(f'/content/knowledge_for_router/{output_file}', 'w') as file:
            file.write(section)

        print(f"Создан файл '{output_file}' с разделом '{section_name}'.")
```

```python
# Замените 'input_file.md' на путь к вашему входному файлу
input_file = '/content/KIA-GPT1/knowledge/database.md'
output_folder = '/content/knowledge_for_router'
!mkdir '/content/knowledge_for_router'

split_file(input_file)
```
